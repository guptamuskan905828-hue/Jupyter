```python
def Customer_Type(M):
    if pd.isna(M):
        return "Unknown"
    elif M>30000:
        return "High Value"
    else:
        return "Regular"
```

```python
#25 Create a new column Customer_Type (High Value / Regular).
df["Customer_Type"]=df["Purchase_Amount"].apply(Customer_Type)
df.head(15)
```
